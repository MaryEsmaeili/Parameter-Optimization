```python
import pandas as pd

def split_location(df):
    """
    Splits the 'source.location' column into 'Side' and 'Lobe' columns.
    """
    df[['Side', 'Lobe']] = df['source.location'].str.split(expand=True, n=1)
    return df

def encode_gender(df):
    """
    Encodes the 'characteristics.tag.gender' column as binary (1 for Male, 0 for Female).
    """
    df['gender_encoded'] = df['characteristics.tag.gender'].map({'M': 1, 'F': 0})
    return df

def create_interaction_feature(df):
    """
    Creates an interaction feature combining 'characteristics.tag.gender' and 'Lobe'.
    """
    df['gender_location'] = df['characteristics.tag.gender'] + '_' + df['Lobe']
    return df

def calculate_sample_frequency(df):
    """
    Adds a frequency count for each 'sample.name'.
    """
    df['sample_frequency'] = df['sample.name'].map(df['sample.name'].value_counts())
    return df

# Example usage:
lung_metadata_df = split_location(lung_metadata_df)
lung_metadata_df = encode_gender(lung_metadata_df)
lung_metadata_df = create_interaction_feature(lung_metadata_df)
lung_metadata_df = calculate_sample_frequency(lung_metadata_df)
print(lung_metadata_df)
```

```python
import numpy as np
from sklearn.decomposition import PCA

def add_statistical_aggregates(df):
    """
    Adds mean, median, and standard deviation of gene expression for each sample.
    """
    df['mean_expression'] = df.iloc[:, 1:].mean(axis=1)
    df['median_expression'] = df.iloc[:, 1:].median(axis=1)
    df['std_expression'] = df.iloc[:, 1:].std(axis=1)
    return df

def apply_pca(df, n_components=5):
    """
    Applies PCA to the gene expression data (excluding the 'ID_REF' column).
    Adds PCA components as new features.
    """
    expression_values = df.iloc[:, 1:]
    pca = PCA(n_components=n_components)
    pca_components = pca.fit_transform(expression_values)
    pca_df = pd.DataFrame(pca_components, columns=[f'PCA_{i+1}' for i in range(pca.n_components_)])
    df = pd.concat([df.reset_index(drop=True), pca_df], axis=1)
    return df

def add_ratio_and_difference(df, gene1, gene2):
    """
    Adds ratio and difference features for two specified gene columns.
    """
    df[f'ratio_{gene1}_{gene2}'] = df[gene1] / df[gene2]
    df[f'diff_{gene1}_{gene2}'] = df[gene1] - df[gene2]
    return df

# Example usage:
gene_expression_df = add_statistical_aggregates(gene_expression_df)
gene_expression_df = apply_pca(gene_expression_df, n_components=5)
gene_expression_df = add_ratio_and_difference(gene_expression_df, 'GSM1416528', 'GSM1416529')
print(gene_expression_df)
```

```python
def combine_aggregates_with_metadata(gene_df, metadata_df):
    """
    Aggregates mean and standard deviation of gene expression per sample and merges with metadata.
    """
    # Aggregate gene expression statistics per sample
    sample_aggregates = gene_df.iloc[:, 1:].agg(['mean', 'std']).transpose()
    sample_aggregates.columns = [f'sample_{stat}' for stat in sample_aggregates.columns]
    
    # Merge with metadata on 'sample.name' (adjust if the column name is different)
    combined_df = metadata_df.join(sample_aggregates, on='sample.name', how='left')
    return combined_df

# Example usage:
combined_df = combine_aggregates_with_metadata(gene_expression_df, lung_metadata_df)
print(combined_df)
```

```python
def missing_data_percentage(df):
    """
    Calculate the percentage of missing (NaN) values in each column and overall in the DataFrame.
    
    Parameters:
    df (DataFrame): The DataFrame to analyze.

    Returns:
    DataFrame: A DataFrame with columns 'Column' and 'NaN_Percentage' for each column in df.
    float: The overall percentage of NaN values in the entire DataFrame.
    """
    # Calculate the percentage of NaNs in each column
    column_nan_percentage = (df.isna().mean() * 100).sort_values(ascending=False)
    
    # Calculate the overall percentage of NaNs in the entire DataFrame
    total_nan_percentage = df.isna().mean().mean() * 100
    
    # Create a summary DataFrame for column-wise NaN percentage
    missing_data_df = pd.DataFrame({
        'Column': column_nan_percentage.index,
        'NaN_Percentage': column_nan_percentage.values
    })
    
    print("Overall percentage of NaN values in the DataFrame: {:.2f}%".format(total_nan_percentage))
    return missing_data_df, total_nan_percentage

# Example usage:
missing_data_df, overall_nan_percentage = missing_data_percentage(combined_df)
print(missing_data_df)
```

```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

def tree_based_feature_importance(X, y, top_n=10):
    """
    Selects top features based on importance from a RandomForestClassifier.
    
    Parameters:
    X (DataFrame): The feature matrix.
    y (Series): The target variable.
    top_n (int): Number of top features to select.

    Returns:
    DataFrame: DataFrame with top features and their importance scores.
    """
    model = RandomForestClassifier()
    model.fit(X, y)
    # Get feature importance
    feature_importances = pd.DataFrame({
        'feature': X.columns,
        'importance': model.feature_importances_
    })
    # Sort by importance and select top_n features
    top_features = feature_importances.sort_values(by='importance', ascending=False).head(top_n)
    return top_features

# Example usage:
top_features = tree_based_feature_importance(X, y, top_n=10)
print(top_features)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

def plot_correlation_matrix(df, threshold=0.8, figsize=(10, 8), cmap="coolwarm", annotate=True):
    """
    Plots the correlation matrix of numeric features and identifies highly correlated pairs.
    
    Parameters:
    df (DataFrame): The feature DataFrame.
    threshold (float): Threshold to highlight pairs with high correlation.
    figsize (tuple): Figure size for the plot.
    cmap (str): Color map for the heatmap.
    annotate (bool): Whether to annotate correlation values.

    Returns:
    list of tuples: Pairs of features with correlation above the threshold.
    """
    # Select only numeric columns for correlation analysis
    numeric_df = df.select_dtypes(include=['number']).copy()

    # Drop columns with near-zero variance to avoid meaningless correlations
    numeric_df = numeric_df.loc[:, numeric_df.var() > 1e-5]

    # Compute the correlation matrix
    corr_matrix = numeric_df.corr()

    # Plot the correlation matrix heatmap
    plt.figure(figsize=figsize)
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # Mask to hide the upper triangle
    sns.heatmap(corr_matrix, mask=mask, cmap=cmap, annot=annotate, fmt=".2f", 
                cbar_kws={"shrink": .8}, vmin=-1, vmax=1, linewidths=.5)

    plt.title("Correlation Matrix of Numeric Features")
    plt.show()

    # Identify pairs of features with high correlation
    high_corr_pairs = [
        (col1, col2) 
        for col1 in corr_matrix.columns 
        for col2 in corr_matrix.columns 
        if col1 != col2 and abs(corr_matrix.loc[col1, col2]) > threshold
    ]
    return high_corr_pairs

# Example usage:
high_corr_pairs = plot_correlation_matrix(combined_df, threshold=0.8, figsize=(12, 10), cmap="coolwarm")
print("Highly correlated pairs:", high_corr_pairs)
```

```python
from sklearn.decomposition import PCA
import pandas as pd

def apply_pca(X, n_components=10):
    """
    Reduces dimensionality of features using PCA.
    
    Parameters:
    X (DataFrame): The feature matrix.
    n_components (int): Number of principal components to keep.

    Returns:
    DataFrame: DataFrame of the top principal components.
    """
    pca = PCA(n_components=n_components)
    X_pca = pca.fit_transform(X)
    pca_df = pd.DataFrame(X_pca, columns=[f'PCA_{i+1}' for i in range(n_components)])
    explained_variance = pca.explained_variance_ratio_
    print("Explained Variance by PCA components:", explained_variance)
    return pca_df

# Example usage:from sklearn.decomposition import PCA
import pandas as pd

def apply_pca(X, n_components=None, variance_ratio=None):
    """
    Reduces dimensionality of features using PCA.
    
    Parameters:
    X (DataFrame): The feature matrix.
    n_components (int, optional): Number of principal components to keep.
    variance_ratio (float, optional): Explained variance ratio to retain (between 0 and 1).

    Returns:
    DataFrame: DataFrame of the top principal components.
    """
    if n_components is None and variance_ratio is not None:
        # Use variance_ratio to determine n_components
        pca = PCA(n_components=variance_ratio)
    else:
        # Ensure n_components is within valid range
        n_components = min(n_components or X.shape[1], X.shape[1])
        pca = PCA(n_components=n_components)
    
    X_pca = pca.fit_transform(X)
    pca_df = pd.DataFrame(X_pca, columns=[f'PCA_{i+1}' for i in range(pca.n_components_)])
    explained_variance = pca.explained_variance_ratio_
    print("Explained Variance by PCA components:", explained_variance)
    return pca_df

# Example usage:
X_pca = apply_pca(X, variance_ratio=0.95)
print(X_pca.head())
```

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

def recursive_feature_elimination(X, y, n_features=10):
    """
    Selects top features using Recursive Feature Elimination (RFE).
    
    Parameters:
    X (DataFrame): The feature matrix.
    y (Series): The target variable.
    n_features (int): Number of features to select.

    Returns:
    list: Names of the selected features.
    """
    model = LogisticRegression(max_iter=1000)
    rfe = RFE(estimator=model, n_features_to_select=n_features)
    rfe.fit(X, y)
    selected_features = X.columns[rfe.support_]
    return selected_features

# Example usage:
selected_features = recursive_feature_elimination(X, y, n_features=10)
print("Selected Features:", selected_features)
```

- **Detecting Overfitting/Underfitting**: A learning curve can help you understand whether your model is overfitting (performing well on training data but poorly on validation data) or underfitting (performing poorly on both training and validation data)
- **ROC**: The ROC curve shows how well your model distinguishes between classes. It helps in selecting the optimal threshold for classification decisions, particularly when the cost of false positives and false negatives differs significantly. ROC curves are often used to compare models. 
- **Confusion matrix**: From the confusion matrix, you can derive other important metrics like precision, recall, F1-score, and specificity, which give a better understanding of how your model is performing across different classes

## Part D. Ensemble learning
Ensemble learning is a powerful machine learning technique that combines the predictions of multiple models to improve overall performance, robustness, and accuracy. Rather than relying on a single model, ensemble learning methods aggregate the results of several models—often called "weak learners"—to produce a stronger predictive model. The key idea behind ensemble learning is that by combining models, the weaknesses of individual models can be offset, leading to better generalization on unseen data. Popular ensemble ML algorithms are the `Random Forest` and `XGBoost`. 
Here’s an improved version of the introduction:

### voting algorithms
Voting algorithms in ensemble learning combine the predictions of multiple classifiers to make a final decision, typically based on the consensus or weighted agreement among models. Two primary types of voting are commonly used: hard voting, where the final prediction is determined by the majority vote, and soft voting, which uses the weighted average of predicted probabilities to determine the outcome. Several sites explain the hard and soft voting algorithm. A clear explanation can be found on https://www.baeldung.com/cs/hard-vs-soft-voting-classifiers

The sklearn pipeline function is a tool in machine learning that simplifies the workflow by encapsulating all the steps involved in a single object. It offers advantages such as simplicity, reproducibility, efficiency, flexibility, and integration. The Pipeline is built using a list of (key, value) pairs, where the key is a string containing the name you want to give this step and value is an estimator object (the method to be executed).
